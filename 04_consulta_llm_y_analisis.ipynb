# Introduccion

## Objetivo del notebook

En este notebook se integra el sistema RAG con un modelo de lenguaje (LLM) para permitir consultas en lenguaje natural sobre la demanda eléctrica argentina.

Se evalúa:

- La calidad de las respuestas.

- La coherencia entre datos y explicaciones.

- La capacidad del sistema para analizar tendencias, comparaciones y relaciones causales.

Este notebook representa la capa cognitiva del proyecto: donde los datos se convierten en análisis y narrativa explicativa.

📌 Analogía argentina:
Acá es donde el analista humano pregunta y la IA responde con respaldo documental.
